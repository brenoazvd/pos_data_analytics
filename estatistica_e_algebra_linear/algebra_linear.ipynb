```python
#BIBLIOTECA
import numpy as np
```

```python
#definir a matriz dos coeficientes
A = np.array([[6,10],[5,7]])
```

```python
B = np.array([8,19])
```

```python
#Resolver o sistema
solucao = np.linalg.solve(A,B)

print(f"valores de x e y: {solucao}")
```

```python
from sympy import symbols, Eq, solve

#definir as variáveis
x, y = symbols('x y')

#definir as equações
eq1 = Eq(2*x + 3*y, 8)
eq2 = Eq(5*x + 7*y, 19)

#resolver o sistema de equações
solucao_simpy = solve((eq1, eq2), (x, y))
print(f"valores de x e y usando sympy: {solucao_simpy}")
```
