Group By

```python
cereal = spark.sql('''SELECT mfr, \
                   type, \
                   COUNT(*) AS total,
                   SUM(calories) AS total_calories
                FROM  cereal \
                 GROUP BY \
                   mfr, \
                   type ''')
cereal.show()
```

CASE WHEN

```python
cereal = spark.sql(''' SELECT mfr, type,
                (case 
                   when type = 'C' then 'A' 
                   when type = 'H' then 'B'
                   else 'C' 
                   end) AS type_new,
                   
                count(*) AS total,
                sum(calories) as total_calories
            FROM cereal
            GROUP BY
                mfr,
                type ''')
    
cereal.show()
```

Avanced Consults in SQL using PySpark

```python
cereal = spark.sql('''
SELECT mfr,
       type,
       sum(calories) as sum_calories,
       min(calories) as min_calories,
       max(calories) as max_calories,
       avg(calories) as avg_calories,
       count(distinct name) as count_distinct_name,
       count(name) as count_names
FROM cereal
GROUP BY mfr, type
ORDER BY mfr, type
''')

cereal.show()
```

```python
cereal = spark.sql('''
SELECT mfr,
       type,
       sum(calories) as sum_calories,
       min(calories) as min_calories,
       max(calories) as max_calories,
       cast(avg(calories) as decimal(10,2)) as avg_calories,

        sum(vitamins) as sum_vitamins,
       min(vitamins) as min_vitamins,
       max(vitamins) as max_vitamins,
       cast(avg(vitamins) as decimal(10,2)) as avg_vitamins,

       sum(carbo) as sum_carbo,
       min(carbo) as min_carbo,
       max(carbo) as max_carbo,
       cast(avg(carbo) as decimal(10,2)) as avg_carbo,

       count(distinct name) as count_distinct_name,
       count(name) as count_names
FROM cereal
GROUP BY mfr, type
ORDER BY mfr, type
''')

cereal.show()
```

# JOINs no PySpark


Dados de exemplo


```python
from pyspark.sql import functions as F

customers_data = [
    (1, "Ana", "SP"),
    (2, "Bruno", "RJ"),
    (3, "Carla", "MG"),
    (4, "Daniel", None),
]

orders_data = [
    (100, 1, "2024-01-05", 120.00),
    (101, 1, "2024-01-07", 75.50),
    (102, 2, "2024-02-01", 220.00),
    (103, 5, "2024-02-11", 15.00),
]

spend_data = [
    (1, "SP", "A"),
    (2, "RJ", "B"),
    (3, "MG", "C"),
    (5, "BA", "D"),
]

channels_data = [
    ("Online",),
    ("Loja",),
]

df_customers = spark.createDataFrame(
    customers_data, ["customer_id", "customer_name", "state"]
)
df_orders = spark.createDataFrame(
    orders_data, ["order_id", "customer_id", "order_date", "amount"]
)
df_spend = spark.createDataFrame(
    spend_data, ["customer_id", "state", "segment"]
)
df_channels = spark.createDataFrame(
    channels_data, ["channel"]
)

print("Clientes:")
df_customers.show()

print("Pedidos:")
df_orders.show()
```

INNER


```python
df_inner = (
    df_orders.alias("o")
    .join(df_customers.alias("c"), on="customer_id", how="inner")
    .select(
        "o.order_id",
        "o.customer_id",
        "c.customer_name",
        "o.order_date",
        "o.amount",
        "c.state",
    )
)

df_inner.show()
```

LEFT


```python
df_left = (
    df_orders.alias("o")
    .join(df_customers.alias("c"), on="customer_id", how="left")
    .select("o.*", "c.customer_name", "c.state")
)

df_left.show()
```

RIGHT


```python
df_right = (
    df_orders.alias("o")
    .join(df_customers.alias("c"), on="customer_id", how="right")
    .select("c.customer_id", "c.customer_name", "o.order_id", "o.amount")
)

df_right.show()
```

FULL OUTER


```python
df_full = (
    df_orders.alias("o")
    .join(df_customers.alias("c"), on="customer_id", how="outer")
    .select("o.order_id", "c.customer_name", "o.amount", "c.state")
)

df_full.show()
```

LEFT SEMI


```python
df_left_semi = df_customers.join(df_orders, on="customer_id", how="left_semi")
df_left_semi.show()
```

LEFT ANTI


```python
df_left_anti = df_customers.join(df_orders, on="customer_id", how="left_anti")
df_left_anti.show()
```

JOIN com colunas de nomes diferentes


```python
df_customer_ref = df_customers.select(
    F.col("customer_id").alias("id"),
    "customer_name",
    "state",
)

df_diff_cols = (
    df_orders.alias("o")
    .join(df_customer_ref.alias("c"), F.col("o.customer_id") == F.col("c.id"), "inner")
    .select("o.order_id", "o.customer_id", "c.customer_name", "o.amount", "c.state")
)

df_diff_cols.show()
```

JOIN com multiplas chaves


```python
df_multi_key = (
    df_customers.join(df_spend, on=["customer_id", "state"], how="inner")
    .select("customer_id", "customer_name", "state", "segment")
)

df_multi_key.show()
```

CROSS JOIN (produto cartesiano)


```python
spark.conf.set("spark.sql.crossJoin.enabled", "true")

df_cross = df_customers.select("customer_id", "customer_name").crossJoin(df_channels)
df_cross.show()
```

JOIN via Spark SQL


```python
df_customers.createOrReplaceTempView("customers")
df_orders.createOrReplaceTempView("orders")

df_sql = spark.sql(
    """
SELECT
  o.order_id,
  o.customer_id,
  c.customer_name,
  o.amount
FROM orders o
INNER JOIN customers c
  ON o.customer_id = c.customer_id
"""
)

df_sql.show()
```
